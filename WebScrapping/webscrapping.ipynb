```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# Step 1: Generate URLs for all pages
num_pages = 39
base_url = "https://stats.espncricinfo.com/ci/engine/stats/index.html"
url_params = "?class=2;spanmin1=01+Jan+2005;spanval1=span;template=results;type=batting"

links = [f"{base_url}{url_params};page={i}" for i in range(1, num_pages)]

# Step 2: Prepare an empty DataFrame
all_data = pd.DataFrame()

# Step 3: Set request headers to avoid blocking
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"
}

# Step 4: Loop through each URL to scrape the data
for page_num, url in enumerate(links, start=1):
    try:
        # Fetch the webpage content
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, "html.parser")

        # Find all tables with the specified class
        tables = soup.find_all("table", class_="engineTable")
        if len(tables) < 3:
            print(f"Page {page_num}: No valid table found.")
            continue

        # The required table is usually the third one
        table = tables[2]

        # Extract headers if DataFrame is empty
        if all_data.empty:
            column_headers = [th.text.strip() for th in table.find_all('th')]
            all_data = pd.DataFrame(columns=column_headers)

        # Extract data rows
        rows = table.find('tbody').find_all('tr')
        for row in rows:
            cells = row.find_all('td')
            row_data = [cell.text.strip() for cell in cells]

            # Only add rows with the correct number of columns
            if len(row_data) == len(all_data.columns):
                all_data.loc[len(all_data)] = row_data

        print(f"Page {page_num} scraped successfully.")

        # Add a small delay to avoid overwhelming the server
        time.sleep(2)

    except Exception as e:
        print(f"Failed to scrape page {page_num}: {e}")

# Step 5: Display and save the data
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print(all_data)
all_data.to_csv("cricket_stats_2005_onwards.csv", index=False)

print("Data scraping completed and saved to cricket_stats_2005_onwards.csv")
```
